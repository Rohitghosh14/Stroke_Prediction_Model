## 7. Data normalization (numerical & categorical)

- **Categorical → numeric:** Label Encoding

### 7.1 Encode categorical columns

## 9. Feature selection & feature ranking

Two complementary methods:
- **ANOVA F-test** (`SelectKBest`) — statistical, univariate
- **Random Forest importance** — captures non-linear relationships & interactions

### 9.1 ANOVA F-test

```python
X = df_processed.drop(columns=["stroke"])
y = df_processed["stroke"]

selector = SelectKBest(score_func=f_classif, k="all")
selector.fit(X, y)
anova_scores = pd.Series(selector.scores_, index=X.columns).sort_values(ascending=False)
print(anova_scores)
```

```python
plt.figure(figsize=(9, 5.5))
anova_scores.plot(kind="barh", color="teal")
plt.gca().invert_yaxis()
plt.title("Feature Ranking - ANOVA F-test Scores")
plt.xlabel("F-score")
plt.show()
```

### 9.2 Random Forest feature importance

```python
rf = RandomForestClassifier(n_estimators=200, random_state=42, class_weight="balanced")
rf.fit(X, y)
rf_importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)
print(rf_importance)
```

```python
plt.figure(figsize=(9, 5.5))
rf_importance.plot(kind="barh", color="darkorange")
plt.gca().invert_yaxis()
plt.title("Feature Ranking - Random Forest Importance")
plt.xlabel("Importance")
plt.show()

top_features = rf_importance.head(6).index.tolist()
print("Top 6 selected features:", top_features)
```

# Note: Now We will apply Machine Learning Classifiers as per our application requirement on Processed Data (Here it is "stroke_processed.csv"). So in this Example we have complete our data preprocessing and cleaning steps by using sample stroke dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.utils import resample
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    confusion_matrix
)

# 1. Load your dataset (treats 'N/A' text as NaN)
df = pd.read_excel(r"..\data\sample_dataset.xlsx", na_values=["N/A", "NA", ""])

# 2. Drop non-informative 'id' column
if 'id' in df.columns:
    df = df.drop(columns=['id'])

# 3. Handle missing values in 'bmi' using median imputation
df['bmi'] = pd.to_numeric(df['bmi'], errors='coerce')
df['bmi'] = df['bmi'].fillna(df['bmi'].median())

# 4. Encode categorical columns
cat_cols = ['gender', 'ever_married', 'work_type', 'Residence_type', 'smoking_status']
le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col].astype(str))

print("Processed Dataset Shape:", df.shape)
print("Class Distribution:\n", df['stroke'].value_counts())

# 5. Split Features (X) and Target (y)
X = df.drop(columns=['stroke'])
y = df['stroke']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 6. Balance ONLY the Training Set (prevents data leakage while fixing 95:5 imbalance)
train_data = pd.concat([X_train, y_train], axis=1)
majority = train_data[train_data['stroke'] == 0]
minority = train_data[train_data['stroke'] == 1]

minority_upsampled = resample(
    minority,
    replace=True,
    n_samples=len(majority),
    random_state=42
)
train_balanced = pd.concat([majority, minority_upsampled])

X_train_bal = train_balanced.drop(columns=['stroke'])
y_train_bal = train_balanced['stroke']

# 7. Feature Scaling (StandardScaler)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_bal)
X_test_scaled = scaler.transform(X_test)

print(f"\nBalanced Train Shape: {X_train_scaled.shape} | Test Shape: {X_test_scaled.shape}")


```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier

# Define 4 Classical ML Models + 1 ANN (5th Model)
models = {
    "1. Logistic Regression": LogisticRegression(random_state=42, max_iter=1000),
    "2. Random Forest": RandomForestClassifier(n_estimators=150, max_depth=10, random_state=42),
    "3. Support Vector Machine (SVM)": SVC(kernel='rbf', probability=True, random_state=42),
    "4. Gradient Boosting": GradientBoostingClassifier(n_estimators=150, random_state=42),
    "5. Artificial Neural Network (ANN)": MLPClassifier(
        hidden_layer_sizes=(64, 32),
        activation='relu',
        solver='adam',
        max_iter=500,
        random_state=42
    )
}

results = []
conf_matrices = {}
roc_data = {}

for name, model in models.items():
    # Train on balanced, scaled training data
    model.fit(X_train_scaled, y_train_bal)
    
    # Evaluate on original unseen test data
    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)[:, 1]
    
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    auc = roc_auc_score(y_test, y_prob)
    
    results.append({
        "Model": name,
        "Accuracy": round(acc, 4),
        "Precision": round(prec, 4),
        "Recall": round(rec, 4),
        "F1-Score": round(f1, 4),
        "AUC-ROC": round(auc, 4)
    })
    
    conf_matrices[name] = confusion_matrix(y_test, y_pred)
    fpr, tpr, _ = roc_curve(y_test, y_prob)
    roc_data[name] = (fpr, tpr, auc)

results_df = pd.DataFrame(results)
print("=== Performance Metrics Across All 5 Models ===")
display(results_df)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()

for idx, (name, cm) in enumerate(conf_matrices.items()):
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        cbar=False,
        ax=axes[idx],
        xticklabels=['No Stroke (0)', 'Stroke (1)'],
        yticklabels=['No Stroke (0)', 'Stroke (1)']
    )
    axes[idx].set_title(name, fontsize=11, fontweight='bold')
    axes[idx].set_xlabel('Predicted Label')
    axes[idx].set_ylabel('Actual Label')

# Remove the unused 6th subplot box (index 5)
fig.delaxes(axes[5])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

for name, (fpr, tpr, auc_score) in roc_data.items():
    plt.plot(fpr, tpr, lw=2, label=f"{name} (AUC = {auc_score:.4f})")

plt.plot(color='gray', linestyle='-', lw=1.5, label='Random Classifier (AUC = 0.5000)')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (1 - Specificity)', fontsize=11)
plt.ylabel('True Positive Rate (Recall / Sensitivity)', fontsize=11)
plt.title('Comparative AUC-ROC Curves (Stroke Dataset)', fontsize=13, fontweight='bold')
plt.legend(loc='lower right', fontsize=9.5)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Rank models for stroke risk prediction analysis
top_3_df = results_df.sort_values(
    by=["AUC-ROC", "F1-Score", "Recall"], ascending=False
).head(3).reset_index(drop=True)

top_3_df.index = ["Rank 1", "Rank 2", "Rank 3"]

print("=== TOP 3 MODELS FOR STROKE RISK PREDICTION ANALYSIS ===")
display(top_3_df)

# Comparative Bar Chart for Top 3 Models
metrics = ["Accuracy", "Precision", "Recall", "F1-Score", "AUC-ROC"]
ax = top_3_df.set_index("Model")[metrics].plot(
    kind="bar", figsize=(10.5, 5.5), width=0.75, colormap="viridis"
)

plt.title("Top 3 Models: Stroke Risk Prediction Comparative Analysis", fontsize=13, fontweight="bold")
plt.ylabel("Score", fontsize=11)
plt.ylim(0, 1.15)
plt.xticks(rotation=15, ha="right")
plt.legend(loc="upper right", ncol=5, fontsize=9)
plt.grid(axis="y", alpha=0.3)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f", fontsize=8, padding=2)

plt.tight_layout()
plt.show()
```
